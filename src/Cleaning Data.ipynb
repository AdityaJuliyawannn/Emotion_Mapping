```python
unique_emotion = df['emotion'].unique()
print(unique_emotion)
```

```python
def normalize_emotion(text):
    if pd.isna(text):
        return text
    
    # Mapping variasi penulisan
    variations = {
        'Proud': ['Bangga'],
        'Trust': ['Loyalty', 'Faith', 'Percaya', 'loyalty'],
        'Joy': ['Love'],
        'Surprise': ['Terkejut', 'Trkejut', 'Kaget', 'Terkjut', 'terkejut', 'kaget', 'Trekejut'],
        'Sadness' : ['Sad'],
        'Anger' : ['Marah', 'Marh']
    }
    
    # Cek setiap kategori
    for main_category, variants in variations.items():
        if text in variants:
            return main_category
    
    return text  # Return as-is jika tidak match
```

```python
df['emotion_normalized'] = df['emotion'].apply(normalize_emotion)

# Lihat hasil
print("Sebelum normalisasi:")
print(df['emotion'].value_counts())

print("\nSetelah normalisasi:")
print(df['emotion_normalized'].value_counts())
```

```python
# hapus baris dengan link diawali 'https://scontent-cgk2'
df = df[~df['video'].str.startswith("https://scontent-cgk2", na=False)]

# reset index biar rapi
df = df.reset_index(drop=True)
```
